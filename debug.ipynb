```python
from concurrent.futures import ThreadPoolExecutor
import queue
from threading import Lock

executor = ThreadPoolExecutor()
```

```python
def write(text):
    print(text)
    return "salad"

fut = executor.submit(write, "foo")
```

```python
fut.result()
```

```python
class Connection:
    def __init__(self):
        self.lock = Lock()
        
    def write(self, command):
        return executor.submit(self._write, command)

    def interrupt(self, command):
        return executor.submit(self._interrupt, command)

    def _write(self, command):
        print("command:", command)
        return 1

    def _interrupt(self, command):
        print("interrupt:", command)
        return 1

connection = Connection()
```

```python
connection.write("foo").result()
```

```python
connection.interrupt("foo").result()
```
